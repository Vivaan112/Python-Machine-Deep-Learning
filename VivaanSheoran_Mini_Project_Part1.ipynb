```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# 1. Load Dataset
df = pd.read_csv("penguins.csv")

# 2. Initial Data Inspection
print("--- FIRST 5 ROWS ---")
print(df.head())

print("\n--- DATA TYPES & NON-NULL COUNTS ---")
print(df.info())

print("\n--- DESCRIPTIVE STATISTICS ---")
print(df.describe(include="all"))

# 3. Handle Missing Values
df_clean = df.dropna(
    subset=[
        "bill_length_mm",
        "bill_depth_mm",
        "flipper_length_mm",
        "body_mass_g",
    ]
).copy()
df_clean["sex"] = df_clean.groupby("species")["sex"].transform(
    lambda x: x.fillna(x.mode()[0])
)

print("\n--- REMAINING MISSING VALUES AFTER CLEANING ---")
print(df_clean.isnull().sum())
```

```python
# Set Seaborn styling
sns.set_theme(style="whitegrid", palette="Set2")

# Plot 1: Histogram + KDE (Distribution of Body Mass by Species)
plt.figure(figsize=(9, 5))
sns.histplot(
    data=df_clean,
    x="body_mass_g",
    hue="species",
    kde=True,
    element="step",
    alpha=0.6,
)
plt.title("Distribution of Body Mass (g) by Species", fontsize=14, pad=12)
plt.xlabel("Body Mass (grams)", fontsize=11)
plt.ylabel("Penguin Count", fontsize=11)
plt.tight_layout()
plt.show()

# Plot 2: Scatter Plot (Bill Length vs. Bill Depth — Simpson's Paradox)
plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=df_clean,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species",
    style="sex",
    s=80,
    alpha=0.9,
)
plt.title(
    "Bill Length vs. Bill Depth across Species & Sex", fontsize=14, pad=12
)
plt.xlabel("Bill Length (mm)", fontsize=11)
plt.ylabel("Bill Depth (mm)", fontsize=11)
plt.tight_layout()
plt.show()

# Plot 3: Box Plot (Flipper Length Distribution Across Islands and Species)
plt.figure(figsize=(9, 5))
sns.boxplot(
    data=df_clean,
    x="island",
    y="flipper_length_mm",
    hue="species",
    dodge=True,
)
plt.title(
    "Flipper Length Distribution across Islands and Species",
    fontsize=14,
    pad=12,
)
plt.xlabel("Island", fontsize=11)
plt.ylabel("Flipper Length (mm)", fontsize=11)
plt.tight_layout()
plt.show()
```
